# Experiment 1: Artificial Neuron

**Aim:** To understand and implement the basic working of an artificial neuron using Python.

**Theory:**

An artificial neuron (perceptron) is the fundamental unit of a neural network. It receives one or more inputs, multiplies each input by an associated weight, adds a bias term, and passes the resulting value through an activation function to produce an output.

$$output = activation\left(\sum_i (input_i \times weight_i) + bias\right)$$

```python
import numpy as np
```

### Define the Artificial Neuron class

```python
class ArtificialNeuron:
    """A single artificial neuron with configurable weights, bias and activation function."""

    def __init__(self, num_inputs, activation="step"):
        self.weights = np.random.uniform(-1, 1, num_inputs)
        self.bias = 0.0
        self.activation = activation

    def _activate(self, x):
        if self.activation == "step":
            return 1 if x >= 0 else 0
        if self.activation == "sigmoid":
            return 1 / (1 + np.exp(-x))
        if self.activation == "relu":
            return max(0.0, x)
        raise ValueError("Unsupported activation function")

    def forward(self, inputs):
        """Compute the neuron's output for a given input vector."""
        weighted_sum = np.dot(inputs, self.weights) + self.bias
        return self._activate(weighted_sum), weighted_sum
```

### Create a neuron and test it on a single input

```python
np.random.seed(42)

neuron = ArtificialNeuron(num_inputs=3, activation="step")
neuron.weights = np.array([0.5, -0.6, 0.1])
neuron.bias = 0.2

sample_inputs = np.array([1.0, 0.5, -1.5])
output, z = neuron.forward(sample_inputs)

print("Inputs           :", sample_inputs)
print("Weights          :", neuron.weights)
print("Bias             :", neuron.bias)
print("Weighted sum (z) :", round(z, 4))
print("Neuron Output    :", output)
```

### Test the neuron on multiple input samples

```python
test_samples = [
    np.array([1, 1, 1]),
    np.array([0, 0, 0]),
    np.array([-1, 2, 0.5]),
]
for s in test_samples:
    out, z = neuron.forward(s)
    print(f"Input: {s} -> Weighted Sum: {z:.4f} -> Output: {out}")
```

### Conclusion
A single artificial neuron successfully computes a weighted sum of its inputs, adds a bias, and applies a step activation function to produce a binary output — demonstrating the basic building block of a neural network.
