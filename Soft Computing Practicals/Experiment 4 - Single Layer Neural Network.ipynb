# Experiment 4: Single-Layer Neural Network

**Aim:** To create a simple single-layer neural network using Python.

**Theory:**

A single-layer neural network (a Perceptron) maps inputs directly to an output using one layer of weights, a bias, and an activation function. It can be trained using the Perceptron Learning Rule:

$$w = w + \eta (target - prediction)\, x, \qquad b = b + \eta (target - prediction)$$

Here, a single-layer perceptron is trained from scratch (NumPy only) to learn the logical **AND** function.

```python
import numpy as np
```

### Define the Single-Layer Neural Network (Perceptron)

```python
class SingleLayerNN:
    def __init__(self, num_inputs, learning_rate=0.1):
        self.weights = np.zeros(num_inputs)
        self.bias = 0.0
        self.lr = learning_rate

    @staticmethod
    def step(x):
        return 1 if x >= 0 else 0

    def predict(self, x):
        z = np.dot(x, self.weights) + self.bias
        return self.step(z)

    def train(self, X, y, epochs=20):
        for epoch in range(epochs):
            total_error = 0
            for xi, target in zip(X, y):
                prediction = self.predict(xi)
                error = target - prediction
                self.weights += self.lr * error * xi
                self.bias += self.lr * error
                total_error += abs(error)
            print(f"Epoch {epoch + 1:2d}/{epochs} - Total Error: {total_error}")
            if total_error == 0:
                print("Converged early.")
                break
```

### Train the network on the logical AND function

```python
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([0, 0, 0, 1])

model = SingleLayerNN(num_inputs=2, learning_rate=0.1)
print("Training single-layer perceptron on the AND function...\n")
model.train(X, y, epochs=10)

print("\nFinal weights:", model.weights)
print("Final bias   :", model.bias)
```

### Test the trained network

```python
for xi in X:
    print(f"Input: {xi} -> Predicted: {model.predict(xi)}")
```

### Conclusion
The single-layer perceptron successfully learns the linearly-separable AND function within a few epochs, converging when the total error reaches zero.
