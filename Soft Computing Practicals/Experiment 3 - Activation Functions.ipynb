# Experiment 3: Sigmoid, Tanh, and ReLU Activation Functions

**Aim:** To implement and visualize Sigmoid, Tanh, and ReLU activation functions using Python.

**Theory:**

Activation functions introduce non-linearity into a neural network, allowing it to learn complex patterns.

- **Sigmoid**: $f(x) = \dfrac{1}{1+e^{-x}}$, Range: (0, 1)
- **Tanh**: $f(x) = \dfrac{e^{x}-e^{-x}}{e^{x}+e^{-x}}$, Range: (-1, 1)
- **ReLU**: $f(x) = \max(0, x)$, Range: [0, ∞)

```python
import numpy as np
import matplotlib.pyplot as plt
```

### Implement the activation functions

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)

def relu(x):
    return np.maximum(0, x)
```

### Sample values at a few points

```python
sample_points = [-5, -1, 0, 1, 5]
print(f"{'x':>6} | {'Sigmoid':>10} | {'Tanh':>10} | {'ReLU':>10}")
print("-" * 46)
for p in sample_points:
    print(f"{p:6} | {sigmoid(p):10.4f} | {tanh(p):10.4f} | {relu(p):10.4f}")
```

### Visualize all three activation functions

```python
x = np.linspace(-10, 10, 400)
y_sigmoid = sigmoid(x)
y_tanh = tanh(x)
y_relu = relu(x)

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.plot(x, y_sigmoid, color="tab:blue")
plt.title("Sigmoid Activation")
plt.xlabel("x"); plt.ylabel("f(x)"); plt.grid(True)

plt.subplot(1, 3, 2)
plt.plot(x, y_tanh, color="tab:orange")
plt.title("Tanh Activation")
plt.xlabel("x"); plt.ylabel("f(x)"); plt.grid(True)

plt.subplot(1, 3, 3)
plt.plot(x, y_relu, color="tab:green")
plt.title("ReLU Activation")
plt.xlabel("x"); plt.ylabel("f(x)"); plt.grid(True)

plt.tight_layout()
plt.savefig("activation_functions.png", dpi=150)
plt.show()
```

### Conclusion
Sigmoid squashes inputs into (0, 1), Tanh squashes them into (-1, 1) and is zero-centred, while ReLU passes positive values unchanged and clips negative values to zero. These properties directly affect how gradients flow during training.
