# Experiment 7: Backpropagation

**Aim:** To understand and implement the basic concept of Backpropagation for training a neural network.

**Theory:**

Backpropagation is the algorithm used to train multi-layer neural networks. It works in two phases:

1. **Forward Pass** – inputs are passed through the network to produce an output.
2. **Backward Pass** – the error (difference between prediction and target) is propagated backward from the output layer to the input layer using the chain rule of calculus, computing the gradient of the loss with respect to every weight and bias so they can be updated via gradient descent.

This notebook implements backpropagation from scratch (NumPy only) for a small neural network with **one hidden layer**, trained on the XOR problem.

```python
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(a):
    # 'a' is the already-computed sigmoid output: sigmoid'(x) = a * (1 - a)
    return a * (1 - a)
```

### Prepare the XOR dataset and initialise the network

```python
np.random.seed(2)

X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([[0], [1], [1], [0]])

n_input, n_hidden, n_output = 2, 4, 1
learning_rate = 0.5
epochs = 10000

W1 = np.random.uniform(-1, 1, (n_input, n_hidden))
b1 = np.zeros((1, n_hidden))
W2 = np.random.uniform(-1, 1, (n_hidden, n_output))
b2 = np.zeros((1, n_output))

loss_history = []
```

### Train the network using forward + backward passes

```python
for epoch in range(epochs):
    # ---------- FORWARD PASS ----------
    z1 = np.dot(X, W1) + b1
    a1 = sigmoid(z1)               # hidden layer output
    z2 = np.dot(a1, W2) + b2
    a2 = sigmoid(z2)               # final predicted output

    # ---------- LOSS (Mean Squared Error) ----------
    loss = np.mean((y - a2) ** 2)
    loss_history.append(loss)

    # ---------- BACKWARD PASS ----------
    d_a2 = (a2 - y) * sigmoid_derivative(a2)          # dL/dz2
    d_W2 = np.dot(a1.T, d_a2) / len(X)
    d_b2 = np.mean(d_a2, axis=0, keepdims=True)

    d_a1 = np.dot(d_a2, W2.T) * sigmoid_derivative(a1)  # dL/dz1
    d_W1 = np.dot(X.T, d_a1) / len(X)
    d_b1 = np.mean(d_a1, axis=0, keepdims=True)

    # ---------- UPDATE WEIGHTS (Gradient Descent step) ----------
    W2 -= learning_rate * d_W2
    b2 -= learning_rate * d_b2
    W1 -= learning_rate * d_W1
    b1 -= learning_rate * d_b1

    if epoch % 2000 == 0:
        print(f"Epoch {epoch:5d} - Loss: {loss:.6f}")

print(f"\nFinal Loss: {loss_history[-1]:.6f}")
```

### Predictions after training

```python
z1 = np.dot(X, W1) + b1
a1 = sigmoid(z1)
z2 = np.dot(a1, W2) + b2
final_output = sigmoid(z2)
for xi, p, target in zip(X, final_output, y):
    print(f"Input: {xi} -> Predicted: {p[0]:.4f} -> Rounded: {round(p[0])} | Actual: {target[0]}")
```

### Conclusion
Backpropagation successfully trains the two-layer network to solve XOR — a problem that requires a hidden layer, verifying that the error gradients are correctly propagated backward through both layers.
