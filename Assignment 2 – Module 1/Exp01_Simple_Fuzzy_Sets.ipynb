# Experiment 1: Simple Fuzzy Sets and Membership Values

**Aim:** To create and visualize simple fuzzy sets in Python using membership values.

## Theory
A **fuzzy set** A on a universe X is defined by a membership function μ_A(x) ∈ [0, 1].
Unlike a crisp set (where μ is 0 or 1), a fuzzy set allows *partial membership*.

A discrete fuzzy set is written as: `A = {μ(x1)/x1 + μ(x2)/x2 + ...}`

### Step 1: Import libraries

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

### Step 2: Define fuzzy sets using membership values
Universe of discourse: ages of people. We define the fuzzy set **Young** and **Old**.

```python
ages  = [10, 20, 30, 40, 50, 60, 70, 80]
young = [1.0, 0.9, 0.6, 0.3, 0.1, 0.0, 0.0, 0.0]
old   = [0.0, 0.0, 0.1, 0.2, 0.4, 0.7, 0.9, 1.0]

# Store as dictionaries (element -> membership)
Young = dict(zip(ages, young))
Old   = dict(zip(ages, old))
print("Young =", Young)
print("Old   =", Old)
```

### Step 3: Display as a table

```python
print(f"{'Age':>5} | {'mu_Young':>9} | {'mu_Old':>7}")
print("-"*28)
for a in ages:
    print(f"{a:>5} | {Young[a]:>9} | {Old[a]:>7}")
```

### Step 4: Visualize the fuzzy sets

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 4))

# Stem/bar plot for discrete sets
ax[0].bar(np.array(ages)-1.5, young, width=3, label='Young', color='tab:blue')
ax[0].bar(np.array(ages)+1.5, old,   width=3, label='Old',   color='tab:red')
ax[0].set_title('Discrete Fuzzy Sets (Bar)')
ax[0].set_xlabel('Age'); ax[0].set_ylabel('Membership'); ax[0].legend()

# Line plot
ax[1].plot(ages, young, 'o-', label='Young')
ax[1].plot(ages, old,   's-', label='Old')
ax[1].set_title('Discrete Fuzzy Sets (Line)')
ax[1].set_xlabel('Age'); ax[1].set_ylabel('Membership'); ax[1].legend(); ax[1].grid(True)
plt.tight_layout(); plt.show()
```

### Step 5: Compare with a crisp set
A crisp 'Young' set (age ≤ 30) only has memberships 0 or 1.

```python
crisp_young = [1 if a <= 30 else 0 for a in ages]
plt.step(ages, crisp_young, where='mid', label='Crisp Young (age<=30)')
plt.plot(ages, young, 'o-', label='Fuzzy Young')
plt.xlabel('Age'); plt.ylabel('Membership'); plt.title('Crisp vs Fuzzy Set')
plt.legend(); plt.grid(True); plt.show()
```

### Step 6: Basic properties of a fuzzy set

```python
young_arr = np.array(young)
print("Support  (mu > 0)  :", [a for a, m in zip(ages, young) if m > 0])
print("Core     (mu = 1)  :", [a for a, m in zip(ages, young) if m == 1])
print("Height             :", young_arr.max())
print("Alpha-cut (alpha=0.5):", [a for a, m in zip(ages, young) if m >= 0.5])
print("Cardinality (sum)  :", young_arr.sum())
```

## Conclusion
A fuzzy set was created from membership values and visualised. Elements can belong partially to a set, unlike crisp sets. Support, core, height and alpha-cut were computed.
