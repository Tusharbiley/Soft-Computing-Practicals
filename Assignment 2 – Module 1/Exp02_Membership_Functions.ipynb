# Experiment 2: Triangular, Trapezoidal and Gaussian Membership Functions

**Aim:** To create and plot Triangular, Trapezoidal, and Gaussian membership functions using Python and scikit-fuzzy.

## Theory
* **Triangular** `trimf(x, [a, b, c])` : rises from a to peak b, falls to c.
* **Trapezoidal** `trapmf(x, [a, b, c, d])` : rises a→b, flat b→c (μ=1), falls c→d.
* **Gaussian** `gaussmf(x, mean, sigma)` : μ(x) = exp(-(x-mean)² / (2σ²)).

### Step 1: Import libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import skfuzzy as fuzz
%matplotlib inline
```

### Step 2: Define universe of discourse

```python
x = np.arange(0, 101, 1)
```

### Step 3: Create the membership functions

```python
tri   = fuzz.trimf(x, [20, 50, 80])
trap  = fuzz.trapmf(x, [10, 30, 60, 90])
gauss = fuzz.gaussmf(x, 50, 12)
```

### Step 4: Plot each function

```python
fig, ax = plt.subplots(1, 3, figsize=(15, 4))
ax[0].plot(x, tri, 'b', lw=2);   ax[0].set_title('Triangular  [20, 50, 80]')
ax[1].plot(x, trap, 'g', lw=2);  ax[1].set_title('Trapezoidal [10, 30, 60, 90]')
ax[2].plot(x, gauss, 'r', lw=2); ax[2].set_title('Gaussian  mean=50, sigma=12')
for a in ax:
    a.set_xlabel('x'); a.set_ylabel('Membership'); a.set_ylim(-0.05, 1.05); a.grid(True)
plt.tight_layout(); plt.show()
```

### Step 5: Compare all together

```python
plt.figure(figsize=(8, 4))
plt.plot(x, tri,   'b', lw=2, label='Triangular')
plt.plot(x, trap,  'g', lw=2, label='Trapezoidal')
plt.plot(x, gauss, 'r', lw=2, label='Gaussian')
plt.title('Comparison of Membership Functions')
plt.xlabel('x'); plt.ylabel('Membership'); plt.legend(); plt.grid(True); plt.show()
```

### Step 6: Find membership value at a specific point

```python
val = 40
print(f"Membership of x={val}:")
print("  Triangular  :", round(float(fuzz.interp_membership(x, tri, val)), 3))
print("  Trapezoidal :", round(float(fuzz.interp_membership(x, trap, val)), 3))
print("  Gaussian    :", round(float(fuzz.interp_membership(x, gauss, val)), 3))
```

### Step 7: Other useful functions (extra)

```python
sig = fuzz.sigmf(x, 50, 0.2)
gb  = fuzz.gbellmf(x, 15, 3, 50)
plt.plot(x, sig, label='Sigmoid'); plt.plot(x, gb, label='Generalized Bell')
plt.legend(); plt.grid(True); plt.title('Sigmoid and Bell MFs'); plt.show()
```

## Conclusion
Triangular, trapezoidal and Gaussian membership functions were created with scikit-fuzzy and plotted. Triangular/trapezoidal are simple and fast; Gaussian is smooth and differentiable.
