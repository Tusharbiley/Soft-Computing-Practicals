# Experiment 2: Weighted Sum of Inputs and Bias

**Aim:** To implement the weighted sum of inputs and bias used in an artificial neuron using Python.

**Theory:**

Before an activation function is applied, a neuron computes a linear combination of its inputs, weights and a bias term:

$$z = (x_1 w_1 + x_2 w_2 + \dots + x_n w_n) + b$$

This experiment implements this calculation in two ways: (1) manually using a loop, and (2) using NumPy's vectorised dot product, and confirms both give the same result.

```python
import numpy as np
```

### Manual (loop-based) weighted sum

```python
def weighted_sum_manual(inputs, weights, bias):
    """Compute weighted sum using an explicit loop (no NumPy)."""
    total = 0.0
    for x, w in zip(inputs, weights):
        total += x * w
    return total + bias
```

### Vectorised (NumPy dot-product) weighted sum

```python
def weighted_sum_vectorised(inputs, weights, bias):
    """Compute weighted sum using NumPy's dot product."""
    return np.dot(inputs, weights) + bias
```

### Compare both implementations

```python
inputs = [1.5, 2.0, -1.0, 0.5]
weights = [0.4, -0.2, 0.6, 0.9]
bias = 0.3

z_manual = weighted_sum_manual(inputs, weights, bias)
z_vector = weighted_sum_vectorised(np.array(inputs), np.array(weights), bias)

print("Inputs  :", inputs)
print("Weights :", weights)
print("Bias    :", bias)
print(f"\nWeighted Sum (manual loop)      : {z_manual:.4f}")
print(f"Weighted Sum (NumPy dot product): {z_vector:.4f}")
print("Both methods match:", np.isclose(z_manual, z_vector))
```

### Effect of varying the bias

```python
for b in [-1, 0, 1, 2]:
    z = weighted_sum_vectorised(np.array(inputs), np.array(weights), b)
    print(f"bias = {b:2d} -> weighted sum = {z:.4f}")
```

### Conclusion
Both the manual loop and the vectorised NumPy implementation produce identical weighted-sum values, confirming the correctness of the computation. Increasing the bias shifts the weighted sum linearly, as expected.
