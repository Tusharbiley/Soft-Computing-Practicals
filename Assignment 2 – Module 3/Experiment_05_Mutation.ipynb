# Experiment 5: Mutation Operation

**Aim:** To implement the mutation operation for introducing variations in individuals using Python.

**Theory:** Mutation randomly changes genes of an individual to maintain diversity and prevent premature convergence. Bit-flip mutation inverts bits in binary chromosomes; swap mutation exchanges two positions in a permutation; Gaussian mutation adds small random noise to real values.

**Step 0:** Import the required libraries and fix the random seed.

```python
import random, numpy as np
import matplotlib.pyplot as plt
random.seed(42); np.random.seed(42)
```

**Step 1:** Implement bit-flip mutation.

```python
def bit_flip(ind, pm=0.2):
    return [1-g if random.random() < pm else g for g in ind]
```

**Step 2:** Implement swap mutation.

```python
def swap_mutation(ind):
    ind = ind[:]; i, j = random.sample(range(len(ind)), 2)
    ind[i], ind[j] = ind[j], ind[i]; return ind
```

**Step 3:** Implement Gaussian mutation.

```python
def gaussian_mutation(ind, pm=0.5, sigma=0.5):
    return [round(g + random.gauss(0, sigma), 3) if random.random() < pm else g for g in ind]
```

**Step 4:** Apply each mutation and show before/after.

```python
b = [1,0,1,1,0,0,1,0]; print("Original binary     :", b); print("Bit-flip            :", bit_flip(b))
p = [1,2,3,4,5,6];     print("Original permutation:", p); print("Swap mutation       :", swap_mutation(p))
r = [1.0,2.0,3.0,4.0]; print("Original real       :", r); print("Gaussian mutation   :", gaussian_mutation(r))
```
