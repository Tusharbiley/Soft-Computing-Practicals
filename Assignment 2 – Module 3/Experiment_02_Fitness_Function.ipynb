# Experiment 2: Fitness Function in a Genetic Algorithm

**Aim:** To define and calculate a simple fitness function for evaluating individuals in a Genetic Algorithm using Python.

**Theory:** The fitness function gives every individual a numeric score showing how good the solution is; the GA uses it to prefer better individuals. Here the problem is to maximise f(x) = x² for x in [0, 31] using a 5-bit chromosome. OneMax (count of 1s) is also shown.

**Step 0:** Import the required libraries and fix the random seed.

```python
import random, numpy as np
import matplotlib.pyplot as plt
random.seed(42); np.random.seed(42)
```

**Step 1:** Define a function to convert a binary chromosome to an integer.

```python
def bits_to_int(b): return int("".join(map(str,b)), 2)
```

**Step 2:** Define the fitness functions.

```python
def fitness(ind): return bits_to_int(ind) ** 2   # f(x) = x^2
def onemax(ind):  return sum(ind)                # number of 1s
```

**Step 3:** Create a random population of 6 individuals.

```python
pop = [[random.randint(0,1) for _ in range(5)] for _ in range(6)]
```

**Step 4:** Calculate and print the fitness of each individual.

```python
print(f"{'Chromosome':<12}{'x':>4}{'f(x)=x^2':>10}{'OneMax':>8}")
for ind in pop:
    print(f"{''.join(map(str,ind)):<12}{bits_to_int(ind):>4}{fitness(ind):>10}{onemax(ind):>8}")
```

**Step 5:** Find the best individual.

```python
best = max(pop, key=fitness)
print("Best individual:", best, "fitness =", fitness(best))
```
