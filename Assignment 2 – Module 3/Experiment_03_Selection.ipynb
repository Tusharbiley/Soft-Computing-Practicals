# Experiment 3: Selection Methods

**Aim:** To implement basic selection methods for selecting suitable individuals from a population using Python.

**Theory:** Selection chooses parents for the next generation, giving fitter individuals a higher chance. Roulette-wheel selection picks in proportion to fitness; tournament selection picks the best of a few random individuals; rank selection picks according to rank rather than raw fitness.

**Step 0:** Import the required libraries, fix the random seed and define the helper functions needed from earlier experiments.

```python
import random, numpy as np
import matplotlib.pyplot as plt
random.seed(42); np.random.seed(42)

def bits_to_int(b): return int("".join(map(str,b)), 2)
def fitness(ind): return bits_to_int(ind) ** 2   # f(x) = x^2
```

**Step 1:** Create a population and calculate fitness values.

```python
pop  = [[random.randint(0,1) for _ in range(5)] for _ in range(8)]
fits = [fitness(i) for i in pop]
print("Population (x, fitness):", [(bits_to_int(p), f) for p, f in zip(pop, fits)])
```

**Step 2:** Implement roulette-wheel selection.

```python
def roulette(pop, fits, k):
    total = sum(fits); chosen = []
    for _ in range(k):
        r, acc = random.uniform(0,total), 0
        for ind, f in zip(pop, fits):
            acc += f
            if acc >= r:
                chosen.append(ind); break
    return chosen
```

**Step 3:** Implement tournament selection.

```python
def tournament(pop, fits, k, size=3):
    chosen = []
    for _ in range(k):
        idx = random.sample(range(len(pop)), size)
        chosen.append(pop[max(idx, key=lambda i: fits[i])])
    return chosen
```

**Step 4:** Implement rank selection.

```python
def rank_selection(pop, fits, k):
    order = sorted(range(len(pop)), key=lambda i: fits[i])
    ranks = {idx: r+1 for r, idx in enumerate(order)}
    return random.choices(pop, weights=[ranks[i] for i in range(len(pop))], k=k)
```

**Step 5:** Select 4 individuals with each method and display them.

```python
for name, fn in [("Roulette", roulette), ("Tournament", tournament), ("Rank", rank_selection)]:
    sel = fn(pop, fits, 4)
    print(f"{name:<11} selected x-values:", [bits_to_int(s) for s in sel])
```
