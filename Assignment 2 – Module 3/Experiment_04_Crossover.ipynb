# Experiment 4: Crossover Operation

**Aim:** To implement the crossover operation for generating new offspring from selected individuals using Python.

**Theory:** Crossover combines the genes of two parents to produce offspring. In one-point crossover the chromosomes are cut at one point and tails swapped; in two-point crossover the part between two cut points is swapped; in uniform crossover each gene is taken from either parent with equal probability.

**Step 0:** Import the required libraries and fix the random seed.

```python
import random, numpy as np
import matplotlib.pyplot as plt
random.seed(42); np.random.seed(42)
```

**Step 1:** Implement one-point crossover.

```python
def one_point(p1, p2):
    pt = random.randint(1, len(p1)-1)
    return p1[:pt]+p2[pt:], p2[:pt]+p1[pt:], pt
```

**Step 2:** Implement two-point crossover.

```python
def two_point(p1, p2):
    a, b = sorted(random.sample(range(1, len(p1)), 2))
    return p1[:a]+p2[a:b]+p1[b:], p2[:a]+p1[a:b]+p2[b:], (a, b)
```

**Step 3:** Implement uniform crossover.

```python
def uniform(p1, p2, prob=0.5):
    c1, c2 = [], []
    for x, y in zip(p1, p2):
        if random.random() < prob: c1.append(x); c2.append(y)
        else:                      c1.append(y); c2.append(x)
    return c1, c2
```

**Step 4:** Define two parents and apply all three crossovers.

```python
p1 = [1,1,1,1,1,1,1,1]
p2 = [0,0,0,0,0,0,0,0]
print("Parent 1:", p1); print("Parent 2:", p2)
c1, c2, pt = one_point(p1, p2); print(f"One-point (cut={pt}):", c1, c2)
c1, c2, pts = two_point(p1, p2); print(f"Two-point (cuts={pts}):", c1, c2)
c1, c2 = uniform(p1, p2);        print("Uniform:", c1, c2)
```
