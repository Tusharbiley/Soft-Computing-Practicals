# Experiment 1: Population and Encoding Methods

**Aim:** To create and represent a population of individuals using different encoding methods in Python.

**Theory:** Encoding converts a candidate solution into a chromosome that a Genetic Algorithm can process. Common methods are binary encoding (bit strings), real-valued encoding (floats), permutation encoding (orderings, e.g. TSP) and integer encoding. A binary chromosome is decoded back to a real value when its fitness is evaluated.

**Step 0:** Import the required libraries and fix the random seed.

```python
import random, numpy as np
import matplotlib.pyplot as plt
random.seed(42); np.random.seed(42)
```

**Step 1:** Set the population size and create a binary-encoded population.

```python
POP = 5
binary_pop = [[random.randint(0,1) for _ in range(8)] for _ in range(POP)]
```

**Step 2:** Create real-valued, permutation and integer-encoded populations.

```python
real_pop    = [[round(random.uniform(-5,5),2) for _ in range(4)] for _ in range(POP)]
perm_pop    = [random.sample(range(1,7), 6) for _ in range(POP)]
integer_pop = [[random.randint(0,9) for _ in range(6)] for _ in range(POP)]
```

**Step 3:** Display every population.

```python
for name, pop in [("Binary", binary_pop), ("Real-valued", real_pop),
                  ("Permutation", perm_pop), ("Integer", integer_pop)]:
    print(f"--- {name} encoding ---")
    for i, ind in enumerate(pop, 1):
        print(f"Individual {i}: {ind}")
```

**Step 4:** Decode the binary chromosomes into real numbers in the range [0, 10].

```python
def decode(bits, lo=0, hi=10):
    return lo + int("".join(map(str,bits)),2) * (hi-lo) / (2**len(bits)-1)
print("Decoded binary individuals:", [round(decode(b),3) for b in binary_pop])
```
