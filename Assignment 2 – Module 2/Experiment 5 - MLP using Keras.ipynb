# Experiment 5: Multilayer Perceptron (MLP) using Keras/TensorFlow

**Aim:** To create a Multilayer Perceptron (MLP) with input, hidden, and output layers using Keras/TensorFlow.

**Theory:**

An MLP is a feedforward neural network consisting of an input layer, one or more hidden layers, and an output layer. Each layer is fully connected (Dense) to the next. Hidden layers typically use non-linear activations (e.g. ReLU) while the output layer's activation depends on the task (Sigmoid for binary classification, Softmax for multi-class classification).

This notebook builds an MLP for the **XOR problem** (2 inputs → 1 binary output) to demonstrate the architecture. XOR cannot be solved by a single-layer network, which is why a hidden layer is required.

```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

tf.random.set_seed(42)
np.random.seed(42)
```

### Prepare the XOR dataset

```python
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]], dtype="float32")
y = np.array([0, 1, 1, 0], dtype="float32")
```

### Build the MLP: Input(2) → Hidden(4, ReLU) → Hidden(4, ReLU) → Output(1, Sigmoid)

```python
model = keras.Sequential([
    layers.Input(shape=(2,)),
    layers.Dense(4, activation="relu", name="hidden_layer_1"),
    layers.Dense(4, activation="relu", name="hidden_layer_2"),
    layers.Dense(1, activation="sigmoid", name="output_layer"),
])

model.summary()
```

### Compile and train the model

```python
model.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])
history = model.fit(X, y, epochs=500, verbose=0)

print("Final training accuracy:", round(history.history["accuracy"][-1], 4))
print("Final training loss    :", round(history.history["loss"][-1], 4))
```

### Predictions on the XOR inputs

```python
predictions = model.predict(X, verbose=0)
for xi, pred, target in zip(X, predictions, y):
    print(f"Input: {xi} -> Predicted: {pred[0]:.4f} (rounded: {round(pred[0])}) | Actual: {int(target)}")
```

### Conclusion
The MLP with two hidden layers successfully learns the non-linearly-separable XOR function, something a single-layer perceptron cannot do — demonstrating the value of hidden layers.
